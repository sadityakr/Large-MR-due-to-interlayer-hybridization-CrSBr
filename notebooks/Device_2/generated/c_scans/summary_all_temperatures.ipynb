# Summary: MR Ratio Comparison Across All Temperatures (c_scans)

This notebook compares MR ratio vs voltage curves for all temperatures in the c_scans dataset.

## Analysis:
- Load MR data from all temperature CSVs
- Plot all curves together for comparison
- Identify temperature-dependent trends

**Note:** c_scans use different field thresholds:
- I_min region: |H| < 0.25 T
- I_max region: |H| > 2.5 T

## 1. Setup & Imports

```python
# Notebook setup
from scripts.utils import setup_notebook
PROJECT_ROOT, np, pd, plt, Path = setup_notebook()

from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

## 2. Load MR Data from All Temperatures

```python
# Path to TMR CSV files
data_dir = PROJECT_ROOT / "output" / "IV_H_scans" / "dataframes" / "c_scans"

# Find all TMR CSV files
tmr_files = sorted(data_dir.glob("TMR_ratio_vs_V_*K.csv"))

print(f"Found {len(tmr_files)} TMR data files:")
for f in tmr_files:
    print(f"  - {f.name}")

# Load all TMR data
tmr_data = {}

for tmr_file in tmr_files:
    # Extract temperature from filename: TMR_ratio_vs_V_10K.csv
    temp_str = tmr_file.stem.split('_')[-1]  # Get '10K'
    temperature = int(temp_str[:-1])  # Remove 'K' and convert to int
    
    # Load data
    df = pd.read_csv(tmr_file)
    tmr_data[temperature] = df
    
    print(f"Loaded {temperature}K: {len(df)} data points")

print(f"\nTotal temperatures loaded: {len(tmr_data)}")
print(f"Temperatures: {sorted(tmr_data.keys())} K")
```

## 3. Plot MR Ratio vs Voltage - All Temperatures

```python
# Create figure
fig, ax = plt.subplots(figsize=(14, 8))

# Color map for different temperatures
temperatures = sorted(tmr_data.keys())
colors = plt.cm.coolwarm(np.linspace(0, 1, len(temperatures)))

# Plot each temperature
for temp, color in zip(temperatures, colors):
    df = tmr_data[temp]
    
    # Plot TMR ratio
    ax.plot(df['Voltage (V)'], df['TMR_Ratio'], '-', 
            color=color, linewidth=2.5, label=f'{temp}K', alpha=0.8)
    
    # Optionally add error bands (commented out for clarity)
    # ax.fill_between(df['Voltage (V)'], 
    #                  df['TMR_Ratio'] - df['TMR_Error'],
    #                  df['TMR_Ratio'] + df['TMR_Error'],
    #                  color=color, alpha=0.1)

# Formatting
ax.set_xlabel('Voltage (V)', fontsize=16)
ax.set_ylabel('MR Ratio (I_max / I_min)', fontsize=16)
ax.set_title('MR Ratio vs Applied Voltage - All Temperatures (c_scans)', fontsize=18, fontweight='bold')
ax.legend(fontsize=12, loc='best', ncol=2)
ax.grid(True, linestyle='--', alpha=0.4)
ax.set_ylim(bottom=0)

plt.tight_layout()
plt.show()

# Save figure
output_fig = PROJECT_ROOT / "notebooks" / "Device_2" / "c_scans" / "TMR_summary_all_temps.png"
fig.savefig(output_fig, dpi=300, bbox_inches='tight')
print(f"\nFigure saved to: {output_fig}")
```

## 4. Plot with Error Bands for Selected Temperatures

```python
# Select a few temperatures to show with error bands for clarity
selected_temps = [10, 20, 30, 40, 50]  # Adjust based on available data

fig, ax = plt.subplots(figsize=(14, 8))

# Filter to available temperatures
available_selected = [t for t in selected_temps if t in tmr_data]
colors = plt.cm.coolwarm(np.linspace(0, 1, len(available_selected)))

for temp, color in zip(available_selected, colors):
    df = tmr_data[temp]
    
    # Plot TMR ratio with error bands
    ax.plot(df['Voltage (V)'], df['TMR_Ratio'], '-', 
            color=color, linewidth=2.5, label=f'{temp}K', alpha=0.9)
    
    ax.fill_between(df['Voltage (V)'], 
                     df['TMR_Ratio'] - df['TMR_Error'],
                     df['TMR_Ratio'] + df['TMR_Error'],
                     color=color, alpha=0.2)

# Formatting
ax.set_xlabel('Voltage (V)', fontsize=16)
ax.set_ylabel('MR Ratio (I_max / I_min)', fontsize=16)
ax.set_title('MR Ratio vs Applied Voltage - Selected Temperatures with Error Bands (c_scans)', 
             fontsize=18, fontweight='bold')
ax.legend(fontsize=12, loc='best')
ax.grid(True, linestyle='--', alpha=0.4)
ax.set_ylim(bottom=0)

plt.tight_layout()
plt.show()
```

## 5. Maximum MR Ratio vs Temperature

```python
# Extract maximum TMR for each temperature
max_tmr_data = []

for temp in sorted(tmr_data.keys()):
    df = tmr_data[temp]
    max_tmr = df['TMR_Ratio'].max()
    max_tmr_voltage = df.loc[df['TMR_Ratio'].idxmax(), 'Voltage (V)']
    
    max_tmr_data.append({
        'Temperature (K)': temp,
        'Max TMR': max_tmr,
        'Voltage at Max TMR (V)': max_tmr_voltage
    })

df_max_tmr = pd.DataFrame(max_tmr_data)

# Plot
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Left: Max TMR vs Temperature
ax1.plot(df_max_tmr['Temperature (K)'], df_max_tmr['Max TMR'], 
         'o-', color='navy', markersize=8, linewidth=2.5)
ax1.set_xlabel('Temperature (K)', fontsize=14)
ax1.set_ylabel('Maximum MR Ratio', fontsize=14)
ax1.set_title('Maximum MR Ratio vs Temperature', fontsize=16, fontweight='bold')
ax1.grid(True, alpha=0.4)

# Right: Voltage at Max TMR vs Temperature
ax2.plot(df_max_tmr['Temperature (K)'], df_max_tmr['Voltage at Max TMR (V)'], 
         's-', color='darkred', markersize=8, linewidth=2.5)
ax2.set_xlabel('Temperature (K)', fontsize=14)
ax2.set_ylabel('Voltage at Max MR (V)', fontsize=14)
ax2.set_title('Voltage at Maximum MR vs Temperature', fontsize=16, fontweight='bold')
ax2.grid(True, alpha=0.4)
ax2.axhline(0, color='black', linestyle='--', linewidth=1, alpha=0.5)

plt.tight_layout()
plt.show()

# Display table
print("\n" + "="*60)
print("Maximum TMR Summary")
print("="*60)
print(df_max_tmr.to_string(index=False))
print("="*60)
```
